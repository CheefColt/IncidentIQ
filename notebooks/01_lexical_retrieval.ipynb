```python
from incidentiq.search import SearchEngine
DATA_PATH = r"E:\incidentiq\data\processed\logs.parquet"
engine = SearchEngine(DATA_PATH)
```

```python
def show_results(results):

    for result in results:

        print(
            f"{result['rank']:2} | "
            f"{result.get('score', result.get('semantic_score', 0)):.4f} | "
            f"{result['message']}"
        )
```

```python
query = "hardware stopped working"
```

```python
bm25_results = engine.search_bm25(
    query,
    top_k=10
)
```

```python
semantic_results = engine.search_semantic(
    query,
    top_k=10
)

show_results(semantic_results)
```

```python
hybrid_results = engine.search_hybrid(
    query,
    top_k=10
)

show_results(hybrid_results)
```

```python
print("========== BM25 ==========")
show_results(bm25_results)

print("\n======= SEMANTIC =========")
show_results(semantic_results)

print("\n========== RRF ===========")
show_results(hybrid_results)
```
