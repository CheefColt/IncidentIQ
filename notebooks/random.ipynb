```python
from incidentiq.pipeline import IncidentIQ

incidentiq = IncidentIQ("../data/processed/logs.parquet")
```

```python
initial_results = incidentiq.engine.search_hybrid(
    "machine check timeout",
    top_k=3
)

initial_context = incidentiq.context_builder.build_context(
    initial_results
)
```

```python
analysis_result = incidentiq.analyzer.analyze(
    query="machine check timeout",
    context=initial_context,
    patterns=[]
)
```

```python
print(
    "Initial IDs:",
    [item["doc_id"] for item in initial_context["evidence"]]
)
```

```python
print(
    "Tool IDs:",
    [item["doc_id"] for item in analysis_result.tool_evidence]
)
```

```python
from incidentiq.evaluation.grounding import validate_evidence_citations

grounding = validate_evidence_citations(
    analysis=analysis_result.analysis,
    initial_evidence=initial_context["evidence"],
    tool_evidence=analysis_result.tool_evidence,
)

print(grounding)
```
